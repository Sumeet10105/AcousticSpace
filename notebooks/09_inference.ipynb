# Notebook 09: Inference
This notebook loads the exported best-performing model and runs prediction on a single file or batch of audio files.

```python
import os
import json
import pickle
import torch
import pandas as pd
from src.single_predict import predict_single_file
from src.models import ResNetCNN, CRNN, AudioSpectrogramTransformer

print("Inference modules loaded successfully!")
```

```python
# Find best model path
with open('best_model/best_model_config.json', 'r') as f:
    config = json.load(f)

model_name = config['model_name']
print(f"Best model is {model_name.upper()}")

if model_name == 'cnn':
    model = ResNetCNN(num_classes=2, pretrained=False)
    model.load_model('best_model/best_model.pt')
elif model_name == 'crnn':
    model = CRNN(num_classes=2, n_mels=64)
    model.load_model('best_model/best_model.pt')
elif model_name == 'ast':
    model = AudioSpectrogramTransformer(num_classes=2, pretrained=False)
    model.load_model('best_model/best_model.pt')
else:
    model = pickle.load(open('best_model/best_model.pkl', 'rb'))
    
model.eval() if hasattr(model, 'eval') else None
print("Best model weights loaded successfully!")
```

```python
# Test single predict
sample_audio = 'd:/Infotact_Solutions/AcousticSpace/datasets/LA/LA/ASVspoof2019_LA_dev/flac/LA_D_1047731.flac'
if os.path.exists(sample_audio):
    result = predict_single_file(sample_audio, model, device='cpu')
    print("--- Prediction Result ---")
    print("Filepath:", result['filepath'])
    print("Prediction:", result['prediction'])
    print("Confidence:", result['confidence'])
    print("Probability:", result['probability'])
    print("Latency:", result['latency_ms'], "ms")
else:
    print(f"Sample file not found at {sample_audio}")
```
