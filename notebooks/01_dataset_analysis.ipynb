# Notebook 01: Dataset Analysis
This notebook visualizes and inspects the ASVspoof 2019 Logical Access (LA) dataset statistics, class imbalance, speaker distribution, and features.

```python
import os
import json
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import Image, display

# Load stats
stats_path = 'reports/dataset_statistics.json'
with open(stats_path, 'r') as f:
    stats = json.load(f)
print("Dataset Statistics Loaded Successfully!")
print("Sample rate:", stats['train']['sample_rate_stats']['unique'])
print("Train duration stats:", stats['train']['duration_stats'])
```

```python
# Display generated class distribution plot
print("--- Class Distribution across Splits ---")
display(Image(filename='plots/class_distribution.png'))
```

```python
# Display Speaker Distribution
print("--- Speaker Distribution (Train Split) ---")
display(Image(filename='plots/speaker_distribution_train.png'))
```

```python
# Display Attack Systems Distribution
print("--- Spoofing System Distribution (Dev Split) ---")
display(Image(filename='plots/attack_distribution_dev.png'))
```
