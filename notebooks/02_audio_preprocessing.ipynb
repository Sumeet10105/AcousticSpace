# Notebook 02: Audio Preprocessing
This notebook implements the complete audio preprocessing pipeline, validates audio signals, and caches preprocessed waveforms.

```python
import os
import torch
import numpy as np
import pandas as pd
from src.preprocessing import (
    load_audio,
    peak_normalize,
    rms_normalize,
    resample_waveform,
    trim_silence,
    remove_silence,
    highpass_filter,
    spectral_subtraction,
    pad_or_truncate,
    pad_or_truncate_to_duration,
    segment_waveform,
)
from src.data.dataset import AudioSpoofingDataset

print("Preprocessing modules imported successfully!")
```

```python
# Set up directories
raw_dir = 'd:/Infotact_Solutions/AcousticSpace/datasets/LA/LA/ASVspoof2019_LA_dev/flac'
protocol_path = 'd:/Infotact_Solutions/AcousticSpace/datasets/LA/LA/ASVspoof2019_LA_cm_protocols/ASVspoof2019.LA.cm.dev.trl.txt'
processed_dir = 'datasets/processed'
os.makedirs(processed_dir, exist_ok=True)

# Instantiate dataset with a subset of 100 samples for validation
dataset = AudioSpoofingDataset(raw_dir, protocol_path, split='dev', max_samples=100)
print(f"Loaded {len(dataset)} samples for preprocessing demo.")
```

```python
# Define processing pipeline and run on subset
preprocessed_meta = []

for idx in range(len(dataset)):
    filepath = dataset.data[idx]
    label = dataset.labels[idx]
    spk_id = dataset.speaker_ids[idx]
    sys_id = dataset.system_ids[idx]
    
    # 1. Load
    y, sr = load_audio(filepath, target_sr=16000, mono=True)
    
    # 2. Denoise
    y = highpass_filter(y, sample_rate=sr, cutoff=80.0)
    y = spectral_subtraction(y, sample_rate=sr)
    
    # 3. Trim silence
    y = trim_silence(y, threshold_db=-40.0)
    
    # 4. Standardize duration
    y = pad_or_truncate_to_duration(y, sample_rate=sr, duration_seconds=4.0)
    
    # 5. Peak normalize
    y = peak_normalize(y, target_peak=0.9)
    
    # Save preprocessed waveform
    out_filename = os.path.basename(filepath).replace('.flac', '.npy')
    out_path = os.path.join(processed_dir, out_filename)
    np.save(out_path, y.cpu().numpy())
    
    preprocessed_meta.append({
        'speaker_id': spk_id,
        'audio_file': os.path.basename(filepath).replace('.flac', ''),
        'system_id': sys_id,
        'key': 'bonafide' if label == 0 else 'spoof',
        'processed_path': out_path
    })

df_processed = pd.DataFrame(preprocessed_meta)
df_processed.to_csv(os.path.join(processed_dir, 'preprocessed_metadata.csv'), index=False)
print("Finished preprocessing subset! Cached preprocessed files and saved metadata.")
print(df_processed.head())
```
